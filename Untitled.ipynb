```python
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
```

```python
def precomputeAugMatrixCPU(dataset):
    """CPU counterpart of precomputeAugMatrix — identical logic, NumPy arrays."""
    n, d = dataset.shape
    data_f32 = dataset.astype(np.float32)
    sq_norms = np.einsum('ij,ij->i', data_f32, data_f32)   # (n,)
    X_aug = np.empty((n, d + 2), dtype=np.float32)
    X_aug[:, :d]  = data_f32
    X_aug[:, d]   = 1.0
    X_aug[:, d+1] = sq_norms
    return X_aug


def _query_aug_cpu(vecs, d):
    """CPU counterpart of _query_aug."""
    sq = np.einsum('ij,ij->i', vecs, vecs)   # (k,)
    V = np.empty((vecs.shape[0], d + 2), dtype=np.float32)
    V[:, :d]  = -2.0 * vecs
    V[:, d]   = sq
    V[:, d+1] = 1.0
    return V


def batchedEuclideanRobustPruneCPU(sources, dataset, X_aug):
    n, d = dataset.shape
    B = len(sources)
    sources_arr = np.array(sources, dtype=np.int64)

    sv = dataset[sources_arr].astype(np.float32)
    dists_matrix = _query_aug_cpu(sv, d) @ X_aug.T
    np.maximum(dists_matrix, 0.0, out=dists_matrix)

    uncov_mask = np.ones((B, n), dtype=np.bool_)
    uncov_mask[np.arange(B), sources_arr] = False

    neighborhoods = [[] for _ in range(B)]
    active = list(range(B))

    while active:
        act = np.array(active, dtype=np.int64)

        act_uncov = uncov_mask[act]
        act_dists = dists_matrix[act]
        
        masked = np.where(act_uncov, act_dists, np.inf)
        waypoints = np.argmin(masked, axis=1)

        act_uncov[np.arange(len(active)), waypoints] = False

        wp_vecs = dataset[waypoints].astype(np.float32)
        W_aug = _query_aug_cpu(wp_vecs, d)
        D = W_aug @ X_aug.T
        np.maximum(D, 0.0, out=D)
        act_uncov &= ~(D < act_dists)

        uncov_mask[act] = act_uncov

        uncov_counts = np.sum(act_uncov, axis=1)

        for li, gi in enumerate(active):
            neighborhoods[gi].append((int(waypoints[li]), int(uncov_counts[li])))

        still_active = np.any(act_uncov, axis=1)
        active = [gi for li, gi in enumerate(active) if still_active[li]]


    return neighborhoods
```

```python
const = 10

n = [100 * (2 ** i) for i in range(9)]
d = [round(const * np.log2(i)) for i in n]
n, d
```

```python
import collections

coverage = [collections.defaultdict(lambda v=n[i]: v) for i in range(len(n))]

for i in tqdm(range(len(n))):
    X = np.random.choice([-1, 1], (n[i], d[i]), p=[0.5, 0.5])
    X_aug = precomputeAugMatrixCPU(X)
    sources = np.arange(0, X.shape[0])

    for j in range(0, len(sources), 50):
        batch = sources[j : j + 50]
        neighborhoods = batchedEuclideanRobustPruneCPU(batch, X, X_aug)

        for edges in neighborhoods:
            for k in range(len(edges)):
                c = edges[k][1] / X.shape[0]
                coverage[i][k+1] -= c
```

```python
pct99 = []
for i in range(len(n)):
    plt.scatter(
        coverage[i].keys(),
        [val / n[i] for val in coverage[i].values()],
        s=5,
        alpha=1
    )
    for j in coverage[i]:
        if coverage[i][j]/n[i] >= 0.99:
            pct99.append(j)
            break
plt.ylim(0, 1.05)
# plt.xlim(-5, 40) 
pct99
```
